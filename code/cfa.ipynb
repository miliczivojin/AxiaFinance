```python
from utils.financials import safe_get
from utils.formatting import format_value
from variables import COMPANY, COLUMN_MAPPING
```

```python
try:
    last_close = COMPANY.info.get("currentPrice")
    market_cap = COMPANY.info.get("marketCap")
    shares_outstanding = COMPANY.info.get("sharesOutstanding")
    week_52_high = COMPANY.info.get("fiftyTwoWeekHigh")
    week_52_low = COMPANY.info.get("fiftyTwoWeekLow")

    print(f"Last Close ${last_close:,.2f}")
    print(f"Market Cap ${format_value(market_cap)}")
    print(f"Shares Outstanding {format_value(shares_outstanding)}")
    print(f"52-Week High ${week_52_high:,.2f}")
    print(f"52-Week Low ${week_52_low:,.2f}")

except Exception as e:
    print(f"Error while fetching current market snapshot: {e}")
```

```python
try:
    market_cap = COMPANY.info.get("marketCap")

    bal = COMPANY.balance_sheet.T
    inc = COMPANY.income_stmt.T

    total_debt = safe_get(bal, COLUMN_MAPPING["total_debt"])
    minority_interest = safe_get(bal, COLUMN_MAPPING["minority_interest"])
    cash = safe_get(bal, COLUMN_MAPPING["cash_equivalents"])
    preferred_stock = safe_get(bal, COLUMN_MAPPING["preferred_stock"])

    ev = market_cap + total_debt + preferred_stock + minority_interest - cash
    ebitda = safe_get(inc, ["EBITDA"])

    ev_ebitda = ev / ebitda

    print(f"EV/EBITDA: {ev_ebitda.iloc[0]:.2f}x")

except Exception as e:
    print(f"Error while calculating EV/EBITDA: {e}")
    ev_ebitda = None
```
